```python
 class Encapsulated:
    def __init__(self, value):
        self.__value = None
        self.value = value

    # Your code here (@property & @value.setter)
    @property
    def value(self):
      return self.__value
    @value.setter
    def value(self, val):
      if not isinstance(val,(int,float)):
        raise TypeError("value should be a number")
      if val < 0:
        return ValueError("value shuold be bigger than 0")
        self.__value = val

obj = Encapsulated(10)
print(obj.value)
print(obj._Encapsulated__value)
```

## Задание 5

Создайте класс, который использует метакласс, объясните как метаклассы влияют на создание классов в Python, и реализуйте контролируемое изменение класса через метакласс.

```python
class Meta(type):
    def __new__(mcs, name, bases, namespace):
        # Your code here (creates class GoodClass)
        namespace["created_by_meta"] = True
        if "hello" not in namespace:
          def hello(self):
            return f"Hello I'm an object of {name} class"
          namespace["hello"] = hello
        return super().__new__(mcs,name,bases,namespace)

class GoodClass(metaclass=Meta):
    pass

obj = GoodClass()

print(obj.hello())
print(obj.created_by_meta)
```

Реализуйте класс с дескрипторами данных и неданных, объясните разницу между ними и механизм вызова методов get, set, delete, особенно при наследовании.

```python
 class DataDescriptor:
    def __get__(self, instance, owner):
      if instance is None:
        return self
      else:
        return instance.__dict__.get("_data_desc", None)

    def __set__(self, instance, value):
      print(f"Set data_desc = {value}")
      instance.__dict__["_data_desc"] = value

    def __delete__(self, instance):
      print(f"Delete {instance}")
      instance.__dict__.pop("_data_desc", None)


class NonDataDescriptor:
    def __get__(self, instance, owner):
      if instance is None:
        return self
      return instance.__dict__.get("_data_desc", None)

class MyClass2:
    data_desc = DataDescriptor()
    non_data_desc = NonDataDescriptor()

obj = MyClass2()
obj.data_desc = 10
print(obj.data_desc)
print(obj.non_data_desc)
```

```python
class A:
  def method(self):
    print("Method A")

class B(A):
  def method(self):
    print("Method B")
    super().method()

class C(A):
  def method(self):
    print("Method C")
    super().method()

class D(B, C):
  def method(self):
    print("Method D")
    super().method()

d = D()
d.method()
print(D.__mro__)
```

## Задание 8

Реализуйте класс с пользовательскими dunder-методами: new, init, call, del, repr, str, и объясните, как Python вызывает их в цикле жизни объекта.

```python
class LifeCycle:
  def __new__(cls,*args,**kwargs):
    print("__new__: New object created")
    instance = super().__new__(cls)
    return instance

  def __init__(self):
    print("__init__: Object is initialized")
  def __call__(self):
    print("__call___: Object is called")
  def __del__(self):
    print("__del__: Object is going to be deleted")
  def __repr__(self):
    return "LifeCycle()"
  def __str__(self):
    return "Object of LifeCycle class"

obj = LifeCycle()
print(obj)
obj()
del obj
```

## Задание 9

Создайте контекстный менеджер с помощью специальных методов enter и exit, используйте его вместе с классом, в котором присутствуют методы с разграничением прав доступа.

https://habr.com/ru/articles/739326/

```python
class Access:
    # Your code here
    def __init__(self):
      self._secret = "secret data"
      self._access_granted = False

    def __enter__(self):
      print("Access allowed")
      self._access_granted = True
      return self._secret

    def __exit__(self, exc_type, exc_val,exc_tb):
      print("Access closed")
      self._access_granted = False

    @property
    def secret(self):
      if self._access_granted:
        return self._secret
      else:
        return "There is no access"

obj = Access()
print(obj.secret)
with obj as secret_value:
    print(secret_value)
print(obj.secret)
```
